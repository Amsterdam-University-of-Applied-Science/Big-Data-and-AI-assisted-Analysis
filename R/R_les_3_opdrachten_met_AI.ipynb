### Statistische analyse van het verband tussen vertrektijd en aankomstvertraging

Om de statistische significantie van het verband tussen vertrektijd en aankomstvertraging te bepalen, kunnen we een correlatietest uitvoeren en vervolgens een lineaire regressieanalyse. Dit helpt ons te kwantificeren hoe sterk de variabelen gerelateerd zijn en of dit verband statistisch significant is.

```r
# Eerst zorgen we ervoor dat we geen NA-waarden hebben voor de analyse
cleaned_flights <- flights %>%
  filter(!is.na(dep_time), !is.na(arr_delay))

# Correlatietest tussen vertrektijd en aankomstvertraging
cor_test_result <- cor.test(cleaned_flights$dep_time, cleaned_flights$arr_delay)
print(cor_test_result)

# Lineaire regressieanalyse
model <- lm(arr_delay ~ dep_time, data = cleaned_flights)
summary(model)
```

## Samenvatten en groeperen

```r
gemiddelde_vertrekvertraging_2013 <- flights %>%
  filter(year == 2013) %>%
  summarise(gemiddelde_vertraging = mean(dep_delay, na.rm = TRUE))

print(gemiddelde_vertrekvertraging_2013)
```

```r
gemiddelde_vertrekvertraging_per_dag_2013 <- flights %>%
  filter(year == 2013) %>%
  group_by(year, month, day) %>%
  summarise(gemiddelde_vertraging = mean(dep_delay, na.rm = TRUE)) %>%
  arrange(year, month, day)

head(gemiddelde_vertrekvertraging_per_dag_2013)
```

```r
install.packages("zoo")
library(zoo)
```

```r
gemiddelde_vertrekvertraging_per_dag_2013 <- gemiddelde_vertrekvertraging_per_dag_2013 %>%
  mutate(date = make_date(year, month, day)) %>%
  mutate(rolling_avg_vertraging = zoo::rollmean(gemiddelde_vertraging, k = 7, fill = NA, align = "right"))

ggplot(gemiddelde_vertrekvertraging_per_dag_2013, aes(x = date, y = gemiddelde_vertraging)) +
  geom_line(color = 'steelblue', aes(linetype = "Dagelijkse Vertraging")) + # Lijn voor dagelijkse vertraging
  geom_line(aes(y = rolling_avg_vertraging, color = 'red', linetype = "Voortschrijdend Gemiddelde"), linewidth = 1) + # Lijn voor voortschrijdend gemiddelde
  labs(title = 'Gemiddelde Vertrekvertraging per Dag in 2013 met Voortschrijdend Gemiddelde',
       x = 'Datum',
       y = 'Gemiddelde Vertraging (minuten)',
       color = "Legenda",
       linetype = "Legenda") +
  scale_color_manual(values = c('steelblue' = 'steelblue', 'red' = 'red')) +
  theme_minimal()
```

```r
gemiddelde_vertrekvertraging_per_dag_2013 <- gemiddelde_vertrekvertraging_per_dag_2013 %>%
  mutate(day_of_week = wday(date, week_start = 1)) # maandag = 1, zondag = 7

# Correlatietest tussen de dag van de week en de gemiddelde vertraging
cor_dag_vertraging <- cor.test(gemiddelde_vertrekvertraging_per_dag_2013$day_of_week,
                               gemiddelde_vertrekvertraging_per_dag_2013$gemiddelde_vertraging)

print(cor_dag_vertraging)
```

De correlatietest is uitgevoerd en laat een Pearson correlatiecoëfficiënt zien van ongeveer -0.078. Dit duidt op een zeer zwakke, negatieve correlatie tussen de dag van de week en de gemiddelde vertrekvertraging.
<p>
De p-waarde van 0.1385 is hoger dan de gebruikelijke significantiedrempel van 0.05, wat betekent dat dit verband statistisch gezien niet significant is.
<p>
Dit suggereert dat er geen significant lineair verband is tussen de dag van de week en de gemiddelde vertrekvertraging op basis van deze gegevens.
