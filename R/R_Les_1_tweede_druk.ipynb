## We gaan een grafiek maken met ggplot
We beginnen met een lege grafiek:

```r
ggplot(data = penguins)
```

Toevoegen: x- en y-as:

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
)
```

Waar blijven de pinguins?<br>
Die gaan we nu toevoegen:

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point()
```

De waarschuwing zegt het al: er waren 2 pinguins met onvolledige data, die zijn verwijderd.

We hebben verschillende soorten pinguins, kunnen we dat in de grafiek zichtbaar maken?

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g, color = species)
) +
  geom_point()
```

Nog een mooi lijntje om te kijken of er een lineair verband is tussen de x en de y?

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g, color = species)
) +
  geom_point() +
  geom_smooth(method = "lm")
```

Mooi, maar we hebben liever 1 lijn dan 3 lijnen, kan dat ook?

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point(mapping = aes(color = species)) +
  geom_smooth(method = "lm")
```

Nu nog verschillende symbolen voor de 3 verschillende soorten pinguins:

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point(mapping = aes(color = species, shape = species)) +
  geom_smooth(method = "lm")
```

En als laatste alle toeters en bellen toevoegen:<br>
- titel<br>
- subtitel<br>
- omschrijving x- en y-as<br>
- aangepast kleurenpalet, ook te lezen door mensen die kleurenblind zijn

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point(aes(color = species, shape = species)) +
  geom_smooth(method = "lm") +
  labs(
    title = "Body mass and flipper length",
    subtitle = "Dimensions for Adelie, Chinstrap, and Gentoo Penguins",
    x = "Flipper length (mm)", y = "Body mass (g)",
    color = "Species", shape = "Species"
  ) +
  scale_color_colorblind()
```

Kan de R-code ook korter? Ja dat kan. We hebben het nu zo geleerd:

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point()
```

We kunnen de code korter maken door 'data = ' en 'mapping =' weg te laten:

```r
ggplot(penguins, aes(x = flipper_length_mm, y = body_mass_g)) +
  geom_point()
```

# 1.2.5.5
De fout ontstaat omdat geom_point() x- en y-waarden nodig heeft, maar die ontbreken in de code. Je hebt alleen de dataset aangegeven, niet wélke variabelen geplot moeten worden. Door een aesthetic mapping toe te voegen met aes(x = …, y = …) weet ggplot welke variabelen het moet gebruiken en werkt de code wel.

# 1.2.5.6
Het argument na.rm bepaalt of ontbrekende waarden (NA’s) automatisch moeten worden verwijderd. De standaardwaarde is FALSE, wat betekent dat ggplot een waarschuwing geeft wanneer NA’s worden verwijderd. Als je na.rm = TRUE instelt, worden NA’s wel verwijderd maar zonder waarschuwing. Hieronder staat een voorbeeld van een scatterplot met na.rm = TRUE.

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point(na.rm = TRUE)
```

# 1.2.5.7

# 1.2.5.8 (1)
Bill_depth_mm should be mapped on a geom level.

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point(mapping = aes(color = bill_depth_mm)) +
  geom_smooth(method = "loess")
```

# 1.2.5.8 (2)
Or should bill_depth_mm be mapped on a global level? Is er een verschil?

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g, color = bill_depth_mm)
) +
  geom_point() +
  geom_smooth(method = "loess")
```

Bill_depth_mm moet worden gemapt aan de color aesthetic.
Hierdoor wordt de snaveldiepte weergegeven via een kleurverloop, zoals in de legenda rechts.

De mapping color = bill_depth_mm moet op het global level geplaatst worden, dus binnen aes() in de ggplot()-functie.
Dit is nodig omdat zowel de punten (geom_point()) als de smooth-lijn (geom_smooth()) dezelfde kleurmapping gebruiken.
Wanneer je het alleen bij geom_point() zou zetten, wordt de smooth-lijn niet correct gekleurd.

# 1.2.5.9
De code maakt een scatterplot waarbij zowel de punten als de smooth-lijnen worden gekleurd op basis van het eiland waarop een pinguïn is gevonden. Dit resulteert in drie verschillende kleuren punten en drie bijbehorende regressielijnen (één per eiland).
<p>
Omdat de se = FALSE als argument wordt gegeven, mist het vervaagde deel om de lijn heen en is alleen de lijn te zien.
<br>
De se = FALSE in de geom_smooth() functie betekent dat de standard error (standaardfout) banden rond de vloeiende lijn niet worden weergegeven.

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g, color = island)
) +
  geom_point() +
  geom_smooth(se = FALSE)
```

# 1.2.5.10
De twee grafieken zien er niet anders uit. Beide gebruiken dezelfde dataset en dezelfde aesthetic mappings voor zowel de punten als de smooth-lijn. Het enige verschil is dat in de eerste grafiek de mapping op globaal niveau wordt gezet (in ggplot()), terwijl in de tweede grafiek dezelfde mapping bij elke geom afzonderlijk wordt herhaald. Omdat de informatie identiek is, produceren de twee codevoorbeelden precies dezelfde visualisatie.

```r
ggplot(
  data = penguins,
  mapping = aes(x = flipper_length_mm, y = body_mass_g)
) +
  geom_point() +
  geom_smooth()
```

```r
ggplot() +
  geom_point(
    data = penguins,
    mapping = aes(x = flipper_length_mm, y = body_mass_g)
  ) +
  geom_smooth(
    data = penguins,
    mapping = aes(x = flipper_length_mm, y = body_mass_g)
  )
```
