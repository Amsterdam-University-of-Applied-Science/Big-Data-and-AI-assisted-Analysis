```r
library(DBI)
library(RSQLite)
```

```r
# Create an ephemeral in-memory RSQLite database
con <- dbConnect(RSQLite::SQLite(), dbname = ":memory:")
dbListTables(con)
```

```r
dbWriteTable(con, "mtcars", mtcars)
dbListTables(con)
```

```r
dbListFields(con, "mtcars")
dbReadTable(con, "mtcars")
```

```r
# You can fetch all results:
res <- dbSendQuery(con, "SELECT * FROM mtcars WHERE cyl = 4")
dbFetch(res)
dbClearResult(res)
```

```r
# Or a chunk at a time
res <- dbSendQuery(con, "SELECT * FROM mtcars WHERE cyl = 4")
while(!dbHasCompleted(res)){
  chunk <- dbFetch(res, n = 5)
  print(nrow(chunk))
}
```

```r
dbClearResult(res)
```

```r
dbDisconnect(con)
```

# XML-bestanden
Ook XML-data kun je met R inlezen en bewerken. Je gebruikt daarbij het [XML2](https://xml2.r-lib.org/://)-package:

```r
install.packages("xml2")
library(xml2)
```

Daarnaast halen we met een GET-commando informatie van een website. Voor het GET-commando heb je het [httr](https://cran.r-project.org/web/packages/httr/vignettes/quickstart.html)-package nodig:

```r
install.packages("httr")
library(httr)
```

We gaan XML-data lezen van de wikipedia-pagina van [Hadley Wickham](https://en.wikipedia.org/wiki/Hadley_Wickham) (de auteur van ons fijne R-boek). We controleren ook eerst even of de data die we ophalen van het XML-formaat is:

```r
url <- "https://en.wikipedia.org/w/api.php?action=query&titles=Hadley%20Wickham&prop=revisions&rvprop=timestamp%7Cuser%7Ccomment%7Ccontent&rvlimit=5&format=xml&rvdir=newer&rvstart=2015-01-14T17%3A12%3A45Z&rvsection=0"

# Get XML revision history
resp_xml <- GET(url)

# Check response is XML 
http_type(resp_xml)
```

Data-formaat klopt, kunnen we nu de revisie-data (= wie heeft er wanneer aan deze pagina gezeten) gaan inlezen:

```r
# Examine returned text with content()
rev_text <- content(resp_xml, as="text")

# Turn rev_text into an XML document
rev_xml <- read_xml(rev_text)

# Examine the structure of rev_xml
xml_structure(rev_xml)
```

We zien nu de structuur van de data, maar nog niet de data zelf:

```r
# Find all nodes using XPATH "/api/query/pages/page/revisions/rev"
xml_find_all(rev_xml, "/api/query/pages/page/revisions/rev")
```

Bovenstaand zie je de (eerste) 5 users die de pagina gewijzigd hebben (tussen 14 januari 2015 en 6 mei 2015).

Als je de tekst van de wikipedia-pagina wilt zien, gebruik je xml_text. Je ziet de tekst 5x, want er zijn 5 ' nodes' gevonden:

```r
# Find all rev nodes anywhere in document
rev_nodes <- xml_find_all(rev_xml, "//rev")
# Use xml_text() to get text from rev_nodes
xml_text(rev_nodes)
```

Als je wat meer informatie wilt, bijvoorbeeld over de eerste user die de pagina gewijzigd heeft, gebruik je xml_attrs() and xml_attr():

```r
# The first rev node
first_rev_node <- xml_find_first(rev_xml, "//rev")

# Find all attributes with xml_attrs()
xml_attrs(first_rev_node)
```

We kunnen ook de userid's van alle 5 personen die een wijziging hebben gemaakt, laten zien: 

```r
# Find user attribute for all rev nodes
xml_attr(rev_nodes, "user")
```

Kunnen we een functie maken die alle bovenstaande acties in 1x doet?<br>
Jazeker:

```r
get_revision_history <- function(article_title){
  # Get raw revision response
  rev_resp <- resp_xml
  
  # Turn the content() of rev_resp into XML
  rev_xml <- read_xml(content(rev_resp, "text"))
  
  # Find revision nodes
  rev_nodes <- xml_find_all(rev_xml, "//rev")

  # Parse out usernames
  user <- xml_attr(rev_nodes, "user")
  
  # Parse out timestamps
  timestamp <- readr::parse_datetime(xml_attr(rev_nodes, "timestamp"))
  
  # Parse out content
  content <- xml_text(rev_nodes)
  
  # Return data frame 
  data.frame(user = user,
    timestamp = timestamp,
    content = substr(content, 1, 40))
}
```

Even kijken of ie werkt:

```r
# Call function for "Hadley Wickham"
get_revision_history(article_title = "Hadley Wickham")
```
