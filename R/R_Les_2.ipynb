## geom_smooth
In plaats van punten kun je ook een grafiek met een ' vloeiende' lijn maken:

```r
ggplot(data = mpg) + 
  geom_smooth(mapping = aes(x = displ, y = hwy))
```

Om de lijn heen loopt een grijs gebied dat, in de vorm van een betrouwbaarheidsinterval, aangeeft hoeveel de punten in de dataset afwijken van de lijn:<br>
- groot grijs gebied = er zijn veel punten onder of boven de blauwe lijn, het BI is breed<br>
- klein grijs gebied = de meeste punten liggen op of in de buurt van de blauwe lijn, het BI is smal.

Je kunt ook meerdere lijnen in 1 grafiek tekenen:

```r
ggplot(data = mpg) + 
  geom_smooth(mapping = aes(x = displ, y = hwy, linetype = drv))
```

Hier zijn, voor 3 verschillende manieren van aandrijving (four-wheel, front-wheel en rear-wheel), 3 lijnen getekend die het verband tussen motorinhoud en verbruik aangeven.

Je kunt bovenstaande grafiek ook nog met verschillende lijnkleuren maken:

```r
  ggplot(data = mpg) +
  geom_smooth(
    mapping = aes(x = displ, y = hwy, color = drv),
    show.legend = TRUE
  )
```

Als je 2 grafiek-typen wilt combineren, kan dat ook:

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy)) + 
  geom_point() + 
  geom_smooth()
```

Boven zie je nu een combinatie van een punt- en een (vloeiende) lijngrafiek. Als we daar nog met behulp van kleuren het type auto aan toevoegen, krijg je een aardig informatieve grafiek (=infographic):

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy)) + 
  geom_point(mapping = aes(color = class)) + 
  geom_smooth()
```

Als je met de lijn-grafiek alleen het verband wilt laten zien tussen motorinhoud en verbruik voor het type auto 'subcompact', kan dat ook:

```r
ggplot(data = mpg, mapping = aes(x = displ, y = hwy)) + 
  geom_point(mapping = aes(color = class)) + 
  geom_smooth(data = filter(mpg, class == "subcompact"), se = FALSE)
```

## geom_bar
Een veelgebruikt grafiektype is het staafdiagram. Als voorbeeld bij dit grafiektype gebruiken we een dataset met daarin informatie over diamanten:

```r
?diamonds
```

We beginnen met een staafdiagram met op de x-as de kwaliteit (cut) en op de y-as het aantal:

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut))
```

Je hoeft bij dit grafiektype alleen op te geven wat je op de x-as wilt, de 'count' (= het aantal) rekent R zelf standaard voor je uit op basis van de gegeven data.

Als je in plaats van 'count' een andere waarde op de y-as wilt, kun je dat aangeven met de 'after_stat'-variabele:

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut, y=after_stat(prop), group=1))
```

Bovenstaand staafdiagram geeft op de y-as de relatieve hoeveelheid  diamanten (prop = proportion) van een bepaalde kwaliteit.

We kunnen de staafdiagrammen nog wat mooier maken met kleuren:

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut, fill = cut))
```

En net als eerder met de punt- en lijngrafiek, kunnen we extra informatie toevoegen aan het staafdiagram waardoor we een nog wat informatievere grafiek krijgen:  

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut, fill = clarity))
```

Als we op de y-as niet de absolute aantallen maar de relatieve frequentie willen zien, kan dat ook met behulp van de 'position'-variabele:

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut, fill = clarity), position = "fill")
```

En met 'position="dodge" ' krijgen we een soort normaalverdeling per soort diamant:

```r
ggplot(data = diamonds) + 
  geom_bar(mapping = aes(x = cut, fill = clarity), position = "dodge")
```

### stat_summary
Met de stat_summary kun je een grafiek maken met daarin een overzicht van (bijv) de minimum, maximum en de middelste waarde van een bepaalde variabele (in dit voorbeeld: de 'depth' van een diamand):

```r
ggplot(data = diamonds) + 
  stat_summary(
    mapping = aes(x = cut, y = depth),
    fun.min = min,
    fun.max = max,
    fun = median
  )
```

## geom_histogram
Met een histogram kun je de frequentieverdeling van een bepaalde variabele laten zien. We gebruiken wederom de diamantendataset als voorbeeld.<br>

Voor meer informatie over histograms:<br>
https://ggplot2.tidyverse.org/reference/geom_histogram.html#ref-examples

We beginnen met een basisversie, waarbij we het aantal karaat op de x-as zetten:

```r
ggplot(diamonds, aes(carat)) +
  geom_histogram()
```

Met 'binwidth' kun je het aantal klassen op de x-as aangeven. Standaard staat dit op 30, maar misschien is bij deze x-as (het aantal karaat) 1 of 0,5 wel beter:

```r
ggplot(diamonds, aes(carat)) +
  geom_histogram(binwidth = 0.5)
```

Je kunt ook een staafdiagram gebruiken met 'scale_x_binned' om een histogram te maken:

```r
# For histograms with tick marks between each bin, use `geom_bar()` with
# `scale_x_binned()`.
ggplot(diamonds, aes(carat)) +
  geom_bar() +
  scale_x_binned()
```

Met kleur erbij kunnen we meerdere variabelen in 1 histogram laten zien:

```r
# Rather than stacking histograms, it's easier to compare frequency
ggplot(diamonds, aes(price, fill = cut)) +
  geom_histogram(binwidth = 500)
```

Zie boven: op de x-as staat de prijs van een diamant, op de y-as het aantal en de kleuren geven de kwaliteit van de diamant aan.

### geom_freqpoly
Je kunt in plaats van staven ook een frequentiepolygoon maken = de middelste waarden per klasse zijn verbonden met een lijn.

```r
# polygons
ggplot(diamonds, aes(price, colour = cut)) +
  geom_freqpoly(binwidth = 500)
```

Als laatste bij de histogrammen een voorbeeld waarbij we meerdere histogrammen tegelijk maken, maar waarbij de schaal op de x-assen verschilt. De 'binwidth' wordt per histogram apart berekend op basis van de schaalgrootte van de variabele die op de x-as staat.
<p>
We gebruiken de economics_long als dataset.

```r
# You can specify a function for calculating binwidth, which is
# particularly useful when faceting along variables with
# different ranges because the function will be called once per facet
ggplot(economics_long, aes(value)) +
  facet_wrap(~variable, scales = 'free_x') +
  geom_histogram(binwidth = function(x) 2 * IQR(x) / (length(x)^(1/3)))
```

## geom_boxplot
Met een boxplot kun je in 1 grafiek een overzicht maken van minimum, 1e kwartiel, mediaan, 3e kwartiel en maximum van een bepaalde variabele. Hieronder is als voorbeeld het verbruik per type auto genomen: 

```r
ggplot(data = mpg, mapping = aes(x = class, y = hwy)) + 
  geom_boxplot()
```

Je kunt de boxplot ook 'flippen', waarbij de x- en y-as omgedraaid worden:

```r
ggplot(data = mpg, mapping = aes(x = class, y = hwy)) + 
  geom_boxplot() +
  coord_flip()
```

## geom_polygon
Je kunt ook kaarten (en andere figuren) maken met de polygon-geom.
Daarvoor installeren we eerst een extra pakket: maps.

```r
install.packages("maps")
```

Vervolgens lezen we kaartgegevens van Nieuw-Zeeland in en tekenen we de kaart:

```r
nz <- map_data("nz")

ggplot(nz, aes(long, lat, group = group)) +
  geom_polygon(fill = "white", colour = "black")
```

## geom_pie
Er is geen aparte 'pie'-geom, maar we kunnen wel een cirkeldiagram maken via een omweg:<br>
- we maken eerst een staafdiagram<br>
- die 'flippen' we (= omdraaien  x- en y-as)<br>
- met de functie coord_polar maken we van het staafdiagram een cirkeldiagram

Eerst het staafdiagram:

```r
bar <- ggplot(data = diamonds) + 
  geom_bar(
    mapping = aes(x = cut, fill = cut), 
    show.legend = FALSE,
    width = 1
  ) + 
  theme(aspect.ratio = 1) +
  labs(x = NULL, y = NULL)
bar
```

Dan flippen:

```r
bar + coord_flip()
```

En nu het cirkeldiagram:

```r
bar + coord_polar()
```
