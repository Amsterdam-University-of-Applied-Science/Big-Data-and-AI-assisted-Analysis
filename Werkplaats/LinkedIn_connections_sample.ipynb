```python
# Script Cell 1
import pandas as pd
import plotly.express as px

df = pd.read_excel('Connections.xlsx')
df.head()

# check if NaN in column "Company"
print(df[df.isna().Company])

# check for NaN in column "Postion"
print(df[df.isna().Position])

# number of rows in the dataframe (before dropping NaN rows)
print(len(df))

# delete rows where "Company" = Nan or "Position" = Nan
df.dropna(subset=['Company','Position'], inplace=True)

# number of rows in the dataframe (after dropping NaN rows)
print(len(df))
```

```python
# add column "CompanyShort" containing first word of "Company"
CompanyShort = df['Company'].split()[0]
df['CompanyShort'] = CompanyShort
  
# Observe the result
print(df)
```

```python
# Script Cell 2
px.treemap(df, title= 'MyNetwork', path=['Company'], width=1200, height=1200)
```

**Mogelijke foutmeldingen**

*ParserError: Error tokenizing data. C error: Expected 1 fields in line 4, saw 6*

Oplossing 1: Wis het commentaar in het begin van het bestand zodat in de eerste regel van het bestand daadwerkelijk de kolom-labels staan.

Oplossing 2: Zie [How To Fix pandas.parser.CParserError: Error tokenizing data](https://towardsdatascience.com/fix-pandas-parser-error-tokenizing-data-889167292d38)

*ValueError: ('None entries cannot have not-None children' Company                 nan
Position    verpleegkundige
Name: 411, dtype: object)*

Betekenis: In een treemap mag de "stam" (de Company) niet de waarde NaN bevatten.

Oplossing: Het verwijderen van de regels waarin in kolom "Company" de waarde NaN staat.
