### Lijndiagram
A line chart or line graph is a type of chart which displays information as a series of data points called 'markers' connected by straight line segments.
A line chart is often used to visualize a trend in data over intervals of time – a time series – thus the line is often drawn chronologically.

### Intermezzo: draaitabel + grafiek

```python
draaitabel_inkomen_geslacht = data.pivot_table(index='edu', columns='sex', values='income', aggfunc='mean', margins=True)
draaitabel_inkomen_geslacht
```

```python
import plotly.express as px

# Exclude the 'All' aggregate row and column for a clearer visualization
df_plot = draaitabel_inkomen_geslacht.drop('All', axis=0).drop('All', axis=1).reset_index()

# Melt the DataFrame to a long format suitable for Plotly Express
df_melted = df_plot.melt(id_vars='edu', var_name='sex', value_name='mean_income')

# Create a grouped bar chart
fig = px.bar(df_melted, x='edu', y='mean_income', color='sex', barmode='group',
             title='Gemiddeld inkomen per opleidingsniveau en geslacht',
             labels={'edu': 'Opleidingsniveau', 'mean_income': 'Gemiddeld inkomen', 'sex': 'Geslacht'})

fig.show()
```

```python
# Create a pivot table to count occurrences (number of individuals) for each edu/sex combination without margins
pivot_counts = data.pivot_table(index='edu', columns='sex', aggfunc='size', fill_value=0)

# Add an 'All' column (sum of FEMALE and MALE for each education level)
pivot_counts['All'] = pivot_counts['FEMALE'] + pivot_counts['MALE']

# Convert counts to row percentages
draaitabel_percentages = pivot_counts.apply(lambda x: x / x['All'] * 100, axis=1)

# Calculate total counts for the 'All' row
total_sex_counts = data['sex'].value_counts(dropna=False)
# Ensure 'FEMALE' and 'MALE' exist in total_sex_counts, fill with 0 if not
total_sex_counts = total_sex_counts.reindex(['FEMALE', 'MALE'], fill_value=0)

# Calculate overall total
overall_total = total_sex_counts.sum()

# Calculate percentages for the 'All' row
all_row_percentages = (total_sex_counts / overall_total) * 100
all_row_percentages['All'] = 100.0 # The 'All' column for the 'All' row should be 100%

# Add 'All' row to draaitabel_percentages
draaitabel_percentages.loc['All'] = all_row_percentages

# Display the pivot table with percentages
draaitabel_percentages
```

```python
import plotly.express as px

# Exclude the 'All' aggregate row for a clearer visualization of education levels
df_plot_percent = draaitabel_percentages.drop('All', axis=0).reset_index()

# Melt the DataFrame to a long format suitable for Plotly Express
df_melted_percent = df_plot_percent.melt(id_vars='edu', var_name='sex', value_name='percentage')

# Create a grouped bar chart, excluding the 'All' column from the sex categories
fig_percent = px.bar(df_melted_percent[df_melted_percent['sex'] != 'All'],
                     x='edu', y='percentage', color='sex', barmode='group',
                     title='Percentage per opleidingsniveau en geslacht',
                     labels={'edu': 'Opleidingsniveau', 'percentage': 'Percentage', 'sex': 'Geslacht'})

fig_percent.show()
```

### Density plot
If the number of data points gets large, scatter plots will no longer be meaningful. Here methods that visualize densities are more useful. The `hexbin` method for _pandas_ data frames is one powerful approach.

```python
ax = data.plot.hexbin(x='age', y='income',
                         gridsize=30, sharex=False, figsize=(5, 4))
ax.set_xlabel('Leeftijd')
ax.set_ylabel('Inkomen')

plt.tight_layout()
plt.show()
```

The _seaborn_ kdeplot is a two-dimensional extension of the density plot. The calculation of the 2D-density for a full dataset might take several minutes. It could be sufficient to create the visualization with a smaller sample of the dataset. With less data points, creating the graph takes only seconds. While some details may be lost, the overall shape is preserved.
<p>
Here we use the full (small!) dataset.

```python
import seaborn as sns

fig, ax = plt.subplots(figsize=(4, 4))
sns.kdeplot(data, x='age', y='income', ax=ax)
ax.set_xlabel('Leeftijd')
ax.set_ylabel('Inkomen')

plt.tight_layout()
plt.show()
```

### Visualizing Multiple Variables with Facets

```python
def hexbin(x, y, color, **kwargs):
    cmap = sns.light_palette(color, as_cmap=True)
    plt.hexbin(x, y, gridsize=25, cmap=cmap, **kwargs)

g = sns.FacetGrid(data, col='edu', col_wrap=2)
g.map(hexbin, 'age', 'income',
      extent=[0, 100, 0, 25000])
g.set_axis_labels('Leeftijd', 'Inkomen')
g.set_titles('Opleiding {col_name}')

plt.tight_layout()
plt.show()
```

# Correlatiematrix (heatmap)
Correlatie geeft de mate van samenhang tussen twee variabelen weer, ofwel in hoeverre twee variabelen elkaar beïnvloeden. De correlatie wordt uitgedrukt in de correlatiecoëfficiënt. De waarde van de correlatiecoëfficiënt ligt altijd tussen -1 en +1.
<p>
Als de correlatiecoëfficiënt ligt tussen de -0,3 en 0,3 is er geen sprake van correlatie.<br>
Als de correlatiecoëfficiënt ligt tussen de (-0,3 -0,6) of (0,3 0,6) is er sprake van zwakke correlatie.<br>
Als de correlatiecoëfficiënt ligt tussen de (-0,6 -0,9) of (0,6 0,9) is er sprake van matige correlatie.<br>
Als de correlatiecoëfficiënt ligt tussen de (-0,9 -1) of (0,9 1) is er sprake van sterke correlatie.<br>

```python
import seaborn as sns

fig, ax = plt.subplots(figsize=(15,15))         # Sample figsize in inches
sns.heatmap(data.corr(),annot=True, linewidths=.5, ax=ax)
```

### Correlatiematrix met ellipsen
The following code shows the strength of the correlation using ellipses.

```python
import numpy as np
from matplotlib.collections import EllipseCollection
from matplotlib.colors import Normalize

def plot_corr_ellipses(data, figsize=None, **kwargs):
    ''' https://stackoverflow.com/a/34558488 '''
    M = np.array(data)
    if not M.ndim == 2:
        raise ValueError('data must be a 2D array')
    fig, ax = plt.subplots(1, 1, figsize=figsize, subplot_kw={'aspect':'equal'})
    ax.set_xlim(-0.5, M.shape[1] - 0.5)
    ax.set_ylim(-0.5, M.shape[0] - 0.5)
    ax.invert_yaxis()

    # xy locations of each ellipse center
    xy = np.indices(M.shape)[::-1].reshape(2, -1).T

    # set the relative sizes of the major/minor axes according to the strength of
    # the positive/negative correlation
    w = np.ones_like(M).ravel() + 0.01
    h = 1 - np.abs(M).ravel() - 0.01
    a = 45 * np.sign(M).ravel()

    ec = EllipseCollection(widths=w, heights=h, angles=a, units='x', offsets=xy,
                           norm=Normalize(vmin=-1, vmax=1),
                           transOffset=ax.transData, array=M.ravel(), **kwargs)
    ax.add_collection(ec)

    # if data is a DataFrame, use the row/column names as tick labels
    if isinstance(data, pd.DataFrame):
        ax.set_xticks(np.arange(M.shape[1]))
        ax.set_xticklabels(data.columns, rotation=90)
        ax.set_yticks(np.arange(M.shape[0]))
        ax.set_yticklabels(data.index)

    return ec, ax

m, ax = plot_corr_ellipses(data.corr(numeric_only=True), figsize=(10, 10), cmap='bwr_r')
cb = plt.colorbar(m, ax=ax)
cb.set_label('Correlation coefficient')

plt.tight_layout()
plt.show()
```
