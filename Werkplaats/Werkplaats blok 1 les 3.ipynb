# Histogrammen
We kunnen de data ook in de vorm van een histogram bekijken:

```python
data.hist()
```

De histogrammen hierboven waren heel simpel om te maken, maar zijn misschien niet zo mooi. Hier komt de kracht van parameters instellen echt goed tot zijn recht. Als je in de documentatie kijkt, zie je dat er 17 parameters zijn om in te stellen.

Het eerste dat we willen aanpassen is de layout. Het is natuurlijk niet mooi als we histogrammen naast en onder elkaar hebben. Het zou mooier zijn als we 3 rijen hebben met alle histogrammen naast elkaar.

In de documentatie staat een layout parameter die (rijen, kolommen) wil.

Daarnaast is de afbeelding wat te klein. Ook daar is een parameter voor, namelijk figsize.

Als laatste kan het niet zo mooi als de y-as niet overal hetzelfde is. Zelfs daar is een parameter voor, namelijk sharey. Die moet een waarde True of False krijgen. Door als parameter te zetten sharey=True wordt de y-as tussen alle histogrammen gedeeld (in dit voorbeeld kiezen we voor False, want niet alle histogrammen hebben dezelfde y-as).

Zo kunnen we een complexere regel code bouwen die de histogrammen precies maakt zoals wij willen.

```python
data.hist(layout=(3, 5), figsize=(15, 7), sharey=False)
```

Per histogram kun je nu bekijken:<br>
- hoe de verdeling van de data is: symmetrisch, linksscheef of rechtsscheef?<br>
- of er een grote spreiding in de data zit?<br>
- of er uitschieters in de data zitten?

## Skewness (of scheefheid)
Geeft aan of de verdeling symmetrisch, links- of rechtsscheef is.
<p>
Als skewness = 0: verdeling is symmetrisch<br>
Als skewness < 0: verdeling is linksscheef, de staarten aan de linkerkant zijn groter / langer dan bij een symmetrische verdeling<br>
Als skewness > 0: verdeling is rechtsscheef, de staarten aan de rechterkant zijn groter / langer dan bij een symmetrische verdeling<br>
<p>
Zie ook: skewness (wikipedia): https://en.wikipedia.org/wiki/Skewness
<p>
In python te berekenen via het commando skew():

```python
data.skew()
```

## Kurtosis
Geeft aan of de verdeling 'spits' (met veel uitschieters) of 'vlak' (met weinig uitschieters) is.
<p>
Als kurtosis = 0: verdeling is symmetrisch<br>
Als kurtosis =< 3: verdeling heeft weinig uitschieters<br>
Als kurtosis > 3: verdeling heeft veel uitschieters<br>
<p>
Zie ook: kurtosis (wikipedia): https://en.wikipedia.org/wiki/Kurtosis
<p>
In python te berekenen via het commando kurtosis():

```python
import scipy.stats as stats
from scipy.stats import kurtosis

kurtosis(data['age'], nan_policy='omit')
```

We zien dat er bij de inkomens behoorlijk wat uitschieters in de data zitten.
